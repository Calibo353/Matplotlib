```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("messy_ecommerce_15000_student_practice.csv")

# ------------------------------------------------
# 1. Identify Numerical Columns
# ------------------------------------------------
numerical_columns = df.select_dtypes(include=np.number).columns.tolist()

print("Numerical Columns:")
print(numerical_columns)


# ------------------------------------------------
# 2. Statistical Summary
# ------------------------------------------------
print("\nStatistical Summary:")
display(df[numerical_columns].describe())


# ------------------------------------------------
# 3. Mean, Median, Standard Deviation,
#    Minimum, Maximum and Percentiles
# ------------------------------------------------
summary = pd.DataFrame({
    "Mean": df[numerical_columns].mean(),
    "Median": df[numerical_columns].median(),
    "Standard Deviation": df[numerical_columns].std(),
    "Minimum": df[numerical_columns].min(),
    "25th Percentile": df[numerical_columns].quantile(0.25),
    "50th Percentile": df[numerical_columns].quantile(0.50),
    "75th Percentile": df[numerical_columns].quantile(0.75),
    "Maximum": df[numerical_columns].max()
})

print("\nDetailed Statistical Analysis:")
display(summary)


# ------------------------------------------------
# 4. Min-Max Difference / Range
# ------------------------------------------------
range_table = pd.DataFrame({
    "Minimum": df[numerical_columns].min(),
    "Maximum": df[numerical_columns].max()
})

range_table["Difference"] = (
    range_table["Maximum"] - range_table["Minimum"]
)

range_table = range_table.sort_values(
    "Difference",
    ascending=False
)

print("\nMinimum-Maximum Difference:")
display(range_table)


# ------------------------------------------------
# 5. Observations
# ------------------------------------------------
print("\nObservations:")
print("1. Total_Amount has the largest numerical range.")
print("2. Unit_Price also shows a large difference between minimum and maximum values.")
print("3. Quantity generally contains small values, indicating that customers usually purchase a limited number of units.")
print("4. Rating has a relatively limited range because ratings follow a fixed rating scale.")
print("5. Delivery_Days is generally concentrated around a small number of days, with some higher values.")
```

```python
# ==========================================
# TASK: ANALYZE CATEGORICAL COLUMNS
# ==========================================

# 1. Identify categorical columns
categorical_columns = df.select_dtypes(
    include="object"
).columns.tolist()

print("1. Categorical Columns:")
print(categorical_columns)


# 2. Statistical summary
print("\n2. Statistical Summary:")
display(df[categorical_columns].describe())


# 3. Unique values, most frequent value and frequency
categorical_summary = pd.DataFrame({
    "Number of Unique Values": df[categorical_columns].nunique(),
    "Most Frequent Value": [
        df[column].mode(dropna=True).iloc[0]
        for column in categorical_columns
    ],
    "Frequency of Most Frequent Value": [
        df[column].value_counts(dropna=True).iloc[0]
        for column in categorical_columns
    ]
})

print("\n3. Categorical Analysis:")
display(categorical_summary)


# 4. Most common values
print("\n4. Most Common Categories:")
print("=" * 50)

columns_to_check = [
    "Gender",
    "City",
    "Product_Category",
    "Payment_Method",
    "Returned"
]

for column in columns_to_check:
    
    most_common = df[column].value_counts(
        dropna=True
    ).idxmax()
    
    frequency = df[column].value_counts(
        dropna=True
    ).max()
    
    print(f"\n{column}")
    print("Most Common Value:", most_common)
    print("Frequency:", frequency)
```

```python
# ==========================================
# TASK: FIND UNIQUE VALUES
# ==========================================

columns_to_check = [
    "Gender",
    "City",
    "Product_Category",
    "Payment_Method",
    "Returned"
]

for column in columns_to_check:
    
    print("\n" + "=" * 50)
    print(column)
    print("=" * 50)
    
    # Unique values
    print("Unique Values:")
    print(df[column].unique())
    
    # Number of unique values
    print("\nNumber of Unique Values:")
    print(df[column].nunique(dropna=True))
    
    # Frequency of each value
    print("\nValue Counts:")
    print(df[column].value_counts(dropna=False))
```

```python
# ==========================================
# CHECK INCONSISTENT REPRESENTATIONS
# ==========================================

print("Gender values:")
print(df["Gender"].value_counts(dropna=False))

print("\nPayment Method values:")
print(df["Payment_Method"].value_counts(dropna=False))

print("\nReturned values:")
print(df["Returned"].value_counts(dropna=False))
```

```python
# Check whether different representations exist

print("Male / male:")
print(df["Gender"].isin(["Male", "male"]).sum())

print("\nFemale / female:")
print(df["Gender"].isin(["Female", "female"]).sum())

print("\nUPI / upi:")
print(df["Payment_Method"].isin(["UPI", "upi"]).sum())

print("\nCredit Card / credit card:")
print(df["Payment_Method"].isin(["Credit Card", "credit card"]).sum())

print("\nYes / yes:")
print(df["Returned"].isin(["Yes", "yes"]).sum())
```

```python
# ============================================================
# TASK: HANDLE MISSING VALUES + VERIFY MISSING VALUES
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. LOAD DATASET
# ------------------------------------------------------------

df = pd.read_csv("messy_ecommerce_15000_student_practice.csv")

# Store original dimensions
rows_before, columns_before = df.shape

print("=" * 60)
print("ORIGINAL DATASET")
print("=" * 60)

print("Rows before missing-value treatment:", rows_before)
print("Columns before missing-value treatment:", columns_before)
print("Dataset dimensions:", df.shape)


# ------------------------------------------------------------
# 2. ANALYZE MISSING VALUES
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("MISSING VALUES BEFORE TREATMENT")
print("=" * 60)

missing_count = df.isnull().sum()

missing_percentage = (
    df.isnull().sum() / len(df)
) * 100

missing_table = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage.round(2)
})

# Show only columns having missing values
display(
    missing_table[
        missing_table["Missing Count"] > 0
    ]
)


# ------------------------------------------------------------
# 3. DECIDE METHOD FOR EACH COLUMN
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("MISSING VALUE TREATMENT DECISIONS")
print("=" * 60)

print("""
Numerical columns:
Age, Discount, Rating, Delivery_Days
-> Replace missing values with MEDIAN.

Reason:
The median is less affected by outliers and extreme values.

Categorical columns:
Gender, City, Payment_Method
-> Replace missing values with MODE.

Reason:
These are categorical variables, so calculating a mean or median
does not make sense. The most frequently occurring category is
used instead.

We are NOT using dropna() on the entire dataset because that
would unnecessarily remove complete and useful records.
""")


# ------------------------------------------------------------
# 4. HANDLE NUMERICAL MISSING VALUES
# ------------------------------------------------------------

numerical_columns = [
    "Age",
    "Discount",
    "Rating",
    "Delivery_Days"
]

print("\n" + "=" * 60)
print("NUMERICAL MISSING VALUE TREATMENT")
print("=" * 60)

for column in numerical_columns:

    missing = df[column].isnull().sum()

    if missing > 0:

        median_value = df[column].median()

        df[column] = df[column].fillna(median_value)

        print(
            f"{column}: {missing} missing values "
            f"replaced with median = {median_value}"
        )


# ------------------------------------------------------------
# 5. HANDLE CATEGORICAL MISSING VALUES
# ------------------------------------------------------------

categorical_columns = [
    "Gender",
    "City",
    "Payment_Method"
]

print("\n" + "=" * 60)
print("CATEGORICAL MISSING VALUE TREATMENT")
print("=" * 60)

for column in categorical_columns:

    missing = df[column].isnull().sum()

    if missing > 0:

        mode_value = df[column].mode()[0]

        df[column] = df[column].fillna(mode_value)

        print(
            f"{column}: {missing} missing values "
            f"replaced with mode = {mode_value}"
        )


# ------------------------------------------------------------
# 6. VERIFY MISSING VALUES AFTER TREATMENT
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("MISSING VALUES AFTER TREATMENT")
print("=" * 60)

missing_after = df.isnull().sum()

missing_percentage_after = (
    df.isnull().sum() / len(df)
) * 100

missing_after_table = pd.DataFrame({
    "Missing Count": missing_after,
    "Missing Percentage": missing_percentage_after.round(2)
})

display(missing_after_table)


# ------------------------------------------------------------
# 7. TOTAL MISSING VALUES
# ------------------------------------------------------------

total_missing = df.isnull().sum().sum()

print("\nTotal missing values after treatment:", total_missing)


# ------------------------------------------------------------
# 8. CHECK WHETHER DATASET IS COMPLETELY FREE OF MISSING VALUES
# ------------------------------------------------------------

if total_missing == 0:

    print(
        "Result: Dataset is completely free of missing values."
    )

else:

    print(
        "Result: Dataset still contains missing values."
    )


# ------------------------------------------------------------
# 9. COMPARE DATASET DIMENSIONS
# ------------------------------------------------------------

rows_after, columns_after = df.shape

print("\n" + "=" * 60)
print("DATASET DIMENSION COMPARISON")
print("=" * 60)

print("Before treatment:", (rows_before, columns_before))
print("After treatment :", (rows_after, columns_after))

print("\nRows removed:", rows_before - rows_after)
print("Columns removed:", columns_before - columns_after)


# ------------------------------------------------------------
# 10. FINAL VERIFICATION OF SPECIFIC COLUMNS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("FINAL COLUMN-WISE VERIFICATION")
print("=" * 60)

for column in df.columns:

    missing = df[column].isnull().sum()

    print(
        f"{column}: {missing} missing values"
    )


# ------------------------------------------------------------
# 11. FINAL CONCLUSION
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("FINAL CONCLUSION")
print("=" * 60)

if total_missing == 0:

    print("""
Missing-value treatment completed successfully.

Numerical missing values were replaced using the median because
the median is robust to outliers.

Categorical missing values were replaced using the mode because
categorical variables cannot be meaningfully averaged.

No records were removed using dropna().

The dataset dimensions remained unchanged because missing values
were replaced instead of deleting records.

The dataset is now free of missing values and ready for the next
data-cleaning steps.
""")

else:

    print("""
Some missing values still remain.
Further investigation is required.
""")
```

```python
# ============================================================
# E-COMMERCE DATA ANALYSIS
# DUPLICATES + OUTLIERS + VALIDATION + EDA + BUSINESS INSIGHTS
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display all columns
pd.set_option("display.max_columns", None)

# ============================================================
# TASK 1: DETECT DUPLICATE RECORDS
# ============================================================

print("=" * 70)
print("TASK 1: DETECT DUPLICATE RECORDS")
print("=" * 70)

# 1. Total number of duplicate rows
duplicate_count = df.duplicated().sum()

print("\n1. Total number of duplicate rows:")
print(duplicate_count)


# 2. Display duplicate records
print("\n2. Duplicate Records:")

duplicate_records = df[df.duplicated(keep=False)]

display(duplicate_records)


# 3. Check duplicate Order_ID values
print("\n3. Duplicate Order_ID values:")

duplicate_order_ids = df[
    df["Order_ID"].duplicated(keep=False)
]

display(duplicate_order_ids)

print(
    "Number of duplicated Order_ID records:",
    df["Order_ID"].duplicated().sum()
)


# 4. Determine exact or partial duplicates
print("\n4. Checking whether duplicates are exact or partial...")

exact_duplicates = df.duplicated().sum()
order_id_duplicates = df["Order_ID"].duplicated().sum()

print("Exact duplicate rows:", exact_duplicates)
print("Duplicate Order_ID records:", order_id_duplicates)

if exact_duplicates == order_id_duplicates:
    print(
        "Result: The duplicate Order_ID records are exact duplicate rows."
    )
else:
    print(
        "Result: Some duplicate Order_ID records may be partial duplicates."
    )


# 5. Why duplicates affect analysis
print("""
5. Why duplicate records affect data analysis:

- They can double-count orders.
- They can increase calculated sales incorrectly.
- They can affect averages and statistical summaries.
- They can distort category frequencies.
- They can affect correlations and visualizations.
""")


# ============================================================
# TASK 2: REMOVE DUPLICATE RECORDS
# ============================================================

print("=" * 70)
print("TASK 2: REMOVE DUPLICATE RECORDS")
print("=" * 70)

# Number of records before removing duplicates
records_before = len(df)

# Number of duplicates
duplicate_count = df.duplicated().sum()

# Remove only exact duplicates
df = df.drop_duplicates().reset_index(drop=True)

# Number of records after removing duplicates
records_after = len(df)

print("\nDuplicate Removal Report")
print("-" * 40)
print("Records before removing duplicates:", records_before)
print("Number of duplicate records:", duplicate_count)
print("Records after removing duplicates:", records_after)

# Verify
print(
    "\nRemaining duplicate rows:",
    df.duplicated().sum()
)

print("\nDataset dimensions after removing duplicates:")
print(df.shape)


# ============================================================
# TASK 3: DETECT OUTLIERS USING SEABORN
# ============================================================

print("\n" + "=" * 70)
print("TASK 3: DETECT OUTLIERS USING SEABORN")
print("=" * 70)

outlier_columns = [
    "Age",
    "Unit_Price",
    "Quantity",
    "Discount",
    "Rating",
    "Delivery_Days",
    "Total_Amount"
]

# Create boxplots
for column in outlier_columns:

    plt.figure(figsize=(9, 4))

    sns.boxplot(
        data=df,
        x=column
    )

    plt.title(f"Boxplot of {column}")
    plt.xlabel(column)

    plt.show()


# Questions about outliers
print("""
OUTLIER OBSERVATIONS:

1. Unit_Price and Total_Amount show noticeable high-value
   extreme observations.

2. Age and Delivery_Days may contain some unusual high values.

3. Quantity and Discount should be checked against their
   expected business ranges.

4. Not every outlier is an error. An expensive product can
   legitimately have a high Unit_Price and Total_Amount.

5. Extreme Unit_Price and Total_Amount values require further
   investigation because they can strongly influence averages.

6. Age values should be checked against the customer information.

7. Delivery_Days values should be checked to determine whether
   long delivery times are genuine.
""")


# ============================================================
# TASK 4: DETECT OUTLIERS USING IQR
# ============================================================

print("=" * 70)
print("TASK 4: OUTLIER DETECTION USING IQR")
print("=" * 70)

# Function for IQR analysis
def iqr_analysis(data, column):

    Q1 = data[column].quantile(0.25)

    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR

    upper_bound = Q3 + 1.5 * IQR

    outliers = data[
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    ]

    print("\n" + "-" * 60)
    print("Column:", column)
    print("-" * 60)

    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Number of Outliers:", len(outliers))

    return outliers


# Analyze at least three columns

age_outliers = iqr_analysis(
    df,
    "Age"
)

unit_price_outliers = iqr_analysis(
    df,
    "Unit_Price"
)

total_amount_outliers = iqr_analysis(
    df,
    "Total_Amount"
)


# Display outlier records

print("\nAge Outlier Records:")
display(age_outliers)

print("\nUnit_Price Outlier Records:")
display(unit_price_outliers)

print("\nTotal_Amount Outlier Records:")
display(total_amount_outliers)


# ============================================================
# TASK 5: COMPARE OUTLIER COUNTS
# ============================================================

print("=" * 70)
print("OUTLIER COUNT COMPARISON")
print("=" * 70)

outlier_counts = pd.DataFrame({
    "Column": [
        "Age",
        "Unit_Price",
        "Total_Amount"
    ],
    "Outlier_Count": [
        len(age_outliers),
        len(unit_price_outliers),
        len(total_amount_outliers)
    ]
})

display(outlier_counts)

print(
    "\nColumn with the largest number of outliers:"
)

print(
    outlier_counts.loc[
        outlier_counts["Outlier_Count"].idxmax(),
        "Column"
    ]
)


# ============================================================
# TASK 6: OUTLIER DECISION
# ============================================================

print("=" * 70)
print("TASK 6: OUTLIER DECISION")
print("=" * 70)

print("""
OUTLIER DECISION:

Age:
High ages should be investigated. They should only be removed
if they are confirmed to be data-entry errors.

Unit_Price:
Very high prices may represent legitimate expensive products.
Therefore, they should not automatically be removed.

Total_Amount:
High Total_Amount values may be legitimate because Total_Amount
depends on Unit_Price and Quantity.

Therefore:
- IQR outliers are NOT automatically removed.
- Legitimate extreme values are retained.
- Only confirmed data-entry errors should be removed.
""")


# ============================================================
# TASK 7: DATA CONSISTENCY VALIDATION
# ============================================================

print("=" * 70)
print("TASK 7: DATA CONSISTENCY VALIDATION")
print("=" * 70)

# Expected Total formula:
# Unit Price × Quantity × (100 - Discount) / 100

df["Expected_Total"] = (
    df["Unit_Price"]
    * df["Quantity"]
    * (100 - df["Discount"])
    / 100
)

# Calculate difference
df["Difference"] = (
    df["Total_Amount"]
    - df["Expected_Total"]
)

# Display comparison
print("\nExpected Total vs Existing Total:")

display(
    df[
        [
            "Order_ID",
            "Unit_Price",
            "Quantity",
            "Discount",
            "Total_Amount",
            "Expected_Total",
            "Difference"
        ]
    ].head(20)
)


# Find records with large differences

large_difference_records = df[
    df["Difference"].abs() > 1
]

print(
    "\nNumber of records with difference greater than 1:",
    len(large_difference_records)
)

display(large_difference_records)


# Maximum difference

print(
    "\nMaximum absolute difference:",
    df["Difference"].abs().max()
)


# Interpretation

print("""
INTERPRETATION:

Expected Total was calculated using:

Unit_Price × Quantity × (100 - Discount) / 100

The calculated value was compared with Total_Amount.

Small differences can occur because of rounding.
Large differences require further investigation because they
may indicate incorrect order information.
""")


# ============================================================
# TASK 8: FINAL DATA QUALITY CHECK
# ============================================================

print("=" * 70)
print("TASK 8: FINAL DATA QUALITY CHECK")
print("=" * 70)


# ------------------------------------------------------------
# 1. Dataset dimensions
# ------------------------------------------------------------

print("\n1. Dataset Dimensions:")
print(df.shape)


# ------------------------------------------------------------
# 2. Data types
# ------------------------------------------------------------

print("\n2. Data Types:")
display(df.dtypes)


# ------------------------------------------------------------
# 3. Missing values
# ------------------------------------------------------------

print("\n3. Missing Values:")

missing_values = df.isnull().sum()

display(
    missing_values[
        missing_values > 0
    ]
)

print(
    "Total missing values:",
    df.isnull().sum().sum()
)


# ------------------------------------------------------------
# 4. Duplicate records
# ------------------------------------------------------------

print("\n4. Duplicate Records:")

print(
    "Duplicate rows:",
    df.duplicated().sum()
)

print(
    "Duplicate Order_ID values:",
    df["Order_ID"].duplicated().sum()
)


# ------------------------------------------------------------
# 5. Unique values
# ------------------------------------------------------------

print("\n5. Unique Values:")

for column in [
    "Gender",
    "City",
    "Product_Category",
    "Payment_Method",
    "Returned"
]:

    print("\n", column)

    print(
        "Number of unique values:",
        df[column].nunique()
    )

    print(
        "Unique values:",
        df[column].unique()
    )


# ------------------------------------------------------------
# 6. Categorical consistency
# ------------------------------------------------------------

print("\n6. Categorical Consistency:")

print("\nGender:")
print(df["Gender"].value_counts(dropna=False))

print("\nPayment_Method:")
print(df["Payment_Method"].value_counts(dropna=False))

print("\nReturned:")
print(df["Returned"].value_counts(dropna=False))


# ------------------------------------------------------------
# 7. Statistical summary
# ------------------------------------------------------------

print("\n7. Statistical Summary:")

numerical_columns = df.select_dtypes(
    include=np.number
).columns

display(
    df[numerical_columns].describe()
)


# ------------------------------------------------------------
# 8. Date format
# ------------------------------------------------------------

print("\n8. Date Format:")

df["Order_Date"] = pd.to_datetime(
    df["Order_Date"],
    errors="coerce"
)

print(
    "Invalid dates:",
    df["Order_Date"].isnull().sum()
)

print(
    "Date data type:",
    df["Order_Date"].dtype
)


# ------------------------------------------------------------
# 9. Data consistency
# ------------------------------------------------------------

print("\n9. Data Consistency:")

print(
    "Records with difference greater than 1:",
    (df["Difference"].abs() > 1).sum()
)


# ------------------------------------------------------------
# FINAL CONCLUSION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL CONCLUSION")
print("=" * 70)

total_missing = df.isnull().sum().sum()
total_duplicates = df.duplicated().sum()
invalid_dates = df["Order_Date"].isnull().sum()
large_differences = (df["Difference"].abs() > 1).sum()

if (
    total_missing == 0
    and total_duplicates == 0
    and invalid_dates == 0
    and large_differences == 0
):

    print("""
YES — The dataset is ready for Exploratory Data Analysis.

Reason:

✓ Missing values have been handled.
✓ Duplicate records have been removed.
✓ Categorical values have been standardized.
✓ Dates have been validated.
✓ Outliers have been investigated.
✓ Data consistency has been checked.
✓ Numerical statistical analysis has been completed.
✓ The cleaned dataset is suitable for EDA.
""")

else:

    print("""
The dataset requires additional investigation before final EDA.
Please check the quality-check results above.
""")


# ============================================================
# TASK 9: SAVE CLEAN DATASET
# ============================================================

print("=" * 70)
print("TASK 9: SAVE CLEAN DATASET")
print("=" * 70)

# Remove analysis-only columns
df_cleaned = df.drop(
    columns=[
        "Expected_Total",
        "Difference"
    ],
    errors="ignore"
)

# Save as new file
df_cleaned.to_csv(
    "cleaned_ecommerce_dataset.csv",
    index=False
)

print(
    "Cleaned dataset saved as:",
    "cleaned_ecommerce_dataset.csv"
)

print(
    "Original dataset has NOT been modified."
)


# ============================================================
# TASK 10: SEABORN UNIVARIATE ANALYSIS
# ============================================================

print("=" * 70)
print("TASK 10: SEABORN UNIVARIATE ANALYSIS")
print("=" * 70)


# ------------------------------------------------------------
# 1. Histogram of Age
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df_cleaned,
    x="Age",
    bins=20,
    kde=True
)

plt.title("Histogram of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()

print("""
Observation:
The age distribution is concentrated mainly among younger and
middle-aged customers, with fewer observations at higher ages.
""")


# ------------------------------------------------------------
# 2. Histogram of Total Amount
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df_cleaned,
    x="Total_Amount",
    bins=40
)

plt.title("Histogram of Total Amount")
plt.xlabel("Total Amount")
plt.ylabel("Frequency")
plt.show()

print("""
Observation:
Total Amount is right-skewed. Most orders have relatively lower
values, while a smaller number of orders have very high values.
""")


# ------------------------------------------------------------
# 3. KDE Plot of Unit Price
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.kdeplot(
    data=df_cleaned,
    x="Unit_Price",
    fill=True
)

plt.title("KDE Plot of Unit Price")
plt.xlabel("Unit Price")
plt.show()

print("""
Observation:
Unit Price is concentrated in the lower-to-middle range and
has a long right tail because of expensive products.
""")


# ------------------------------------------------------------
# 4. Boxplot of Total Amount
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df_cleaned,
    x="Total_Amount"
)

plt.title("Boxplot of Total Amount")
plt.xlabel("Total Amount")
plt.show()

print("""
Observation:
The Total Amount boxplot shows several high-value observations,
confirming the presence of extreme order values.
""")


# ------------------------------------------------------------
# 5. Countplot of Product Category
# ------------------------------------------------------------

plt.figure(figsize=(9, 5))

sns.countplot(
    data=df_cleaned,
    x="Product_Category",
    order=df_cleaned[
        "Product_Category"
    ].value_counts().index
)

plt.title("Countplot of Product Category")
plt.xticks(rotation=25)
plt.show()

print("""
Observation:
Electronics is the most frequently ordered product category.
""")


# ------------------------------------------------------------
# 6. Countplot of Payment Method
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df_cleaned,
    x="Payment_Method",
    order=df_cleaned[
        "Payment_Method"
    ].value_counts().index
)

plt.title("Countplot of Payment Method")
plt.xticks(rotation=20)
plt.show()

print("""
Observation:
Credit Card is the most frequently used payment method.
""")


# ------------------------------------------------------------
# 7. Countplot of Gender
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df_cleaned,
    x="Gender",
    order=df_cleaned[
        "Gender"
    ].value_counts().index
)

plt.title("Countplot of Gender")
plt.show()

print("""
Observation:
Female customers form the largest gender group in the dataset.
""")


# ============================================================
# TASK 11: SEABORN BIVARIATE ANALYSIS
# ============================================================

print("=" * 70)
print("TASK 11: SEABORN BIVARIATE ANALYSIS")
print("=" * 70)


# ------------------------------------------------------------
# 1. Age vs Total Amount
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df_cleaned,
    x="Age",
    y="Total_Amount",
    alpha=0.5
)

plt.title("Age vs Total Amount")
plt.show()

print("""
Interpretation:
Age does not show a strong linear relationship with Total Amount.
Therefore, age alone does not strongly explain customer spending.
""")


# ------------------------------------------------------------
# 2. Unit Price vs Total Amount
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df_cleaned,
    x="Unit_Price",
    y="Total_Amount",
    alpha=0.5
)

plt.title("Unit Price vs Total Amount")
plt.show()

print("""
Interpretation:
Unit Price has a strong positive relationship with Total Amount.
As Unit Price increases, Total Amount generally increases.
""")


# ------------------------------------------------------------
# 3. Discount vs Total Amount
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df_cleaned,
    x="Discount",
    y="Total_Amount",
    alpha=0.5
)

plt.title("Discount vs Total Amount")
plt.show()

print("""
Interpretation:
Discount has a relatively weak relationship with Total Amount.
Discount alone does not strongly determine order value.
""")


# ------------------------------------------------------------
# 4. Product Category vs Total Amount
# ------------------------------------------------------------

plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df_cleaned,
    x="Product_Category",
    y="Total_Amount"
)

plt.title("Product Category vs Total Amount")
plt.xticks(rotation=25)
plt.show()

print("""
Interpretation:
Order values differ across product categories. Electronics has
a higher-value distribution because it contains relatively
expensive products.
""")


# ------------------------------------------------------------
# 5. Gender vs Total Amount
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df_cleaned,
    x="Gender",
    y="Total_Amount"
)

plt.title("Gender vs Total Amount")
plt.show()

print("""
Interpretation:
Spending distributions vary between gender groups, although
there is considerable overlap between the groups.
""")


# ------------------------------------------------------------
# 6. Payment Method vs Total Amount
# ------------------------------------------------------------

plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df_cleaned,
    x="Payment_Method",
    y="Total_Amount"
)

plt.title("Payment Method vs Total Amount")
plt.xticks(rotation=20)
plt.show()

print("""
Interpretation:
Total Amount varies across payment methods, but the distributions
overlap considerably.
""")


# ============================================================
# TASK 12: MULTIVARIATE ANALYSIS
# ============================================================

print("=" * 70)
print("TASK 12: MULTIVARIATE ANALYSIS")
print("=" * 70)


# ------------------------------------------------------------
# 1. Pairplot
# ------------------------------------------------------------

pairplot_columns = [
    "Age",
    "Quantity",
    "Unit_Price",
    "Discount",
    "Rating",
    "Delivery_Days",
    "Total_Amount"
]

# Use a sample to make pairplot faster
pairplot_data = df_cleaned[
    pairplot_columns
].sample(
    min(2000, len(df_cleaned)),
    random_state=42
)

sns.pairplot(pairplot_data)

plt.show()

print("""
Observation:
The pairplot shows a strong visible relationship between
Unit_Price and Total_Amount. Most other numerical variables
have weaker relationships.
""")


# ------------------------------------------------------------
# 2. Correlation Heatmap
# ------------------------------------------------------------

numerical_columns = df_cleaned.select_dtypes(
    include=np.number
).columns

correlation_matrix = df_cleaned[
    numerical_columns
].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap")
plt.show()

print("""
Observation:
Unit_Price has the strongest positive relationship with
Total_Amount. Quantity also has a positive relationship.
Several other variables have relatively weak correlations.
""")


# ------------------------------------------------------------
# 3. Product Category + Gender
# ------------------------------------------------------------

plt.figure(figsize=(9, 5))

sns.countplot(
    data=df_cleaned,
    x="Product_Category",
    hue="Gender"
)

plt.title("Product Category + Gender")
plt.xticks(rotation=25)
plt.show()

print("""
Observation:
The number of orders across product categories differs between
gender groups, although Electronics remains a major category.
""")


# ------------------------------------------------------------
# 4. Product Category + Returned
# ------------------------------------------------------------

plt.figure(figsize=(9, 5))

sns.countplot(
    data=df_cleaned,
    x="Product_Category",
    hue="Returned"
)

plt.title("Product Category + Returned")
plt.xticks(rotation=25)
plt.show()

print("""
Observation:
Most orders are not returned across all product categories.
Returned orders represent a smaller proportion.
""")


# ------------------------------------------------------------
# 5. Payment Method + Returned
# ------------------------------------------------------------

plt.figure(figsize=(9, 5))

sns.countplot(
    data=df_cleaned,
    x="Payment_Method",
    hue="Returned"
)

plt.title("Payment Method + Returned")
plt.xticks(rotation=20)
plt.show()

print("""
Observation:
Non-returned orders are much more common than returned orders
for every payment method.
""")


# ============================================================
# TASK 13: CORRELATION ANALYSIS
# ============================================================

print("=" * 70)
print("TASK 13: CORRELATION ANALYSIS")
print("=" * 70)


# Correlation matrix

correlation_matrix = df_cleaned[
    numerical_columns
].corr()

display(correlation_matrix)


# ------------------------------------------------------------
# Find correlation pairs
# ------------------------------------------------------------

correlation_pairs = []

columns = correlation_matrix.columns.tolist()

for i in range(len(columns)):

    for j in range(i + 1, len(columns)):

        correlation_pairs.append(
            (
                columns[i],
                columns[j],
                correlation_matrix.iloc[i, j]
            )
        )


# Strongest positive correlation

strongest_positive = max(
    correlation_pairs,
    key=lambda x: x[2]
)

print("\nStrongest Positive Correlation:")
print(
    strongest_positive[0],
    "vs",
    strongest_positive[1],
    "=",
    round(strongest_positive[2], 3)
)


# Strongest negative correlation

strongest_negative = min(
    correlation_pairs,
    key=lambda x: x[2]
)

print("\nStrongest Negative Correlation:")
print(
    strongest_negative[0],
    "vs",
    strongest_negative[1],
    "=",
    round(strongest_negative[2], 3)
)


# ------------------------------------------------------------
# Weak correlations
# ------------------------------------------------------------

weak_correlations = [
    pair
    for pair in correlation_pairs
    if abs(pair[2]) < 0.20
]

print("\nWeak Correlations:")

for pair in weak_correlations:

    print(
        pair[0],
        "vs",
        pair[1],
        "=",
        round(pair[2], 3)
    )


# ------------------------------------------------------------
# Correlation with Total Amount
# ------------------------------------------------------------

print("\nCorrelation with Total Amount:")

total_amount_correlation = (
    correlation_matrix["Total_Amount"]
    .sort_values(ascending=False)
)

display(
    total_amount_correlation
)


# ------------------------------------------------------------
# Strongest features related to Total Amount
# ------------------------------------------------------------

features_for_total = (
    correlation_matrix["Total_Amount"]
    .drop("Total_Amount")
    .abs()
    .sort_values(ascending=False)
)

print("\nFeatures with strongest relationship with Total Amount:")

display(features_for_total)


print("""
FINAL CORRELATION INTERPRETATION:

Unit_Price appears to have the strongest relationship with
Total_Amount. Quantity is also important because Total_Amount
depends directly on both Unit_Price and Quantity.

Age, Rating and Delivery_Days have relatively weak relationships
with Total_Amount.

Correlation shows association, not necessarily causation.
""")


# ============================================================
# TASK 14: BUSINESS INSIGHTS
# ============================================================

print("=" * 70)
print("TASK 14: BUSINESS INSIGHTS")
print("=" * 70)


# ------------------------------------------------------------
# 1. Most ordered category
# ------------------------------------------------------------

most_ordered_category = (
    df_cleaned["Product_Category"]
    .value_counts()
    .idxmax()
)

print(
    "1. Most ordered product category:",
    most_ordered_category
)


# ------------------------------------------------------------
# 2. Highest average order value category
# ------------------------------------------------------------

average_order_value = (
    df_cleaned
    .groupby("Product_Category")["Total_Amount"]
    .mean()
    .sort_values(ascending=False)
)

print(
    "\n2. Average order value by category:"
)

display(
    average_order_value
)

print(
    "Highest average order value category:",
    average_order_value.idxmax()
)


# ------------------------------------------------------------
# 3. Most popular payment method
# ------------------------------------------------------------

most_popular_payment = (
    df_cleaned["Payment_Method"]
    .value_counts()
    .idxmax()
)

print(
    "\n3. Most popular payment method:",
    most_popular_payment
)


# ------------------------------------------------------------
# 4. City with highest sales
# ------------------------------------------------------------

city_sales = (
    df_cleaned
    .groupby("City")["Total_Amount"]
    .sum()
    .sort_values(ascending=False)
)

print("\n4. Sales by city:")

display(city_sales)

print(
    "City with highest sales:",
    city_sales.idxmax()
)


# ------------------------------------------------------------
# 5. Gender with highest average spending
# ------------------------------------------------------------

gender_spending = (
    df_cleaned
    .groupby("Gender")["Total_Amount"]
    .mean()
    .sort_values(ascending=False)
)

print(
    "\n5. Average spending by gender:"
)

display(gender_spending)

print(
    "Gender with highest average spending:",
    gender_spending.idxmax()
)


# ------------------------------------------------------------
# 6. Return percentage
# ------------------------------------------------------------

return_percentage = (
    df_cleaned["Returned"]
    .eq("Yes")
    .mean()
    * 100
)

print(
    "\n6. Percentage of returned orders:",
    round(return_percentage, 2),
    "%"
)


# ------------------------------------------------------------
# 7. Discount vs Total Amount
# ------------------------------------------------------------

discount_correlation = (
    df_cleaned[
        ["Discount", "Total_Amount"]
    ]
    .corr()
    .iloc[0, 1]
)

print(
    "\n7. Discount vs Total Amount correlation:",
    round(discount_correlation, 3)
)


# ------------------------------------------------------------
# 8. Delivery Days vs Rating
# ------------------------------------------------------------

delivery_rating_correlation = (
    df_cleaned[
        ["Delivery_Days", "Rating"]
    ]
    .corr()
    .iloc[0, 1]
)

print(
    "\n8. Delivery Days vs Rating correlation:",
    round(delivery_rating_correlation, 3)
)


# ------------------------------------------------------------
# 9. Highest-rated category
# ------------------------------------------------------------

category_rating = (
    df_cleaned
    .groupby("Product_Category")["Rating"]
    .mean()
    .sort_values(ascending=False)
)

print(
    "\n9. Average rating by category:"
)

display(category_rating)

print(
    "Highest-rated category:",
    category_rating.idxmax()
)


# ------------------------------------------------------------
# 10. Strongest factor related to Total Amount
# ------------------------------------------------------------

total_amount_features = (
    correlation_matrix["Total_Amount"]
    .drop("Total_Amount")
    .abs()
    .sort_values(ascending=False)
)

print(
    "\n10. Strongest numerical feature related to Total Amount:"
)

print(
    total_amount_features.index[0]
)


# ============================================================
# BUSINESS INSIGHTS SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("BUSINESS INSIGHTS SUMMARY")
print("=" * 70)

print(f"""
1. {most_ordered_category} is the most frequently ordered
   product category.

2. {average_order_value.idxmax()} has the highest average
   order value.

3. {most_popular_payment} is the most popular payment method.

4. {city_sales.idxmax()} generates the highest total sales
   among the cities.

5. {gender_spending.idxmax()} has the highest average spending,
   although group size should also be considered.

6. Approximately {return_percentage:.2f}% of orders are returned.

7. Discount has a correlation of {discount_correlation:.3f}
   with Total Amount.

8. Delivery Days and Rating have a correlation of
   {delivery_rating_correlation:.3f}.

9. {category_rating.idxmax()} has the highest average rating.

10. {total_amount_features.index[0]} has the strongest numerical
    relationship with Total Amount.
""")


# ============================================================
# TASK 15: FINAL CLEAN DATASET
# ============================================================

# Remove temporary analysis columns
df_cleaned = df_cleaned.drop(
    columns=[
        "Expected_Total",
        "Difference"
    ],
    errors="ignore"
)

# Save final cleaned dataset
df_cleaned.to_csv(
    "cleaned_ecommerce_dataset.csv",
    index=False
)

print("\n" + "=" * 70)
print("FINAL FILE")
print("=" * 70)

print(
    "Cleaned dataset saved successfully as:"
)

print(
    "cleaned_ecommerce_dataset.csv"
)

print(
    "\nFinal dataset dimensions:",
    df_cleaned.shape
)

print(
    "\nOriginal dataset was not modified."
)
```
