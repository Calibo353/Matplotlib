```python
system_prompt="""
You are a Campus Buddy.

You help engineering students.

Explain concepts clearly.

If you don't know something,
say "I don't know."

Keep answers under 100 words.
Rules (Guardrails):
1. Help only with engineering and academic topics.
2. Explain concepts in simple English.
3. Give examples whenever possible.
4. If you don't know the answer, say "I don't know."
5. Keep answers under 100 words.
6. Do not provide harmful, illegal, or unsafe instructions.
7. Do not generate offensive or abusive content.
8. Protect the user's privacy. Never ask for or store sensitive personal information.
9. If the question is not related to engineering or academics, politely say:
10."Sorry, I can only help with engineering and academic questions."
10. Be friendly, respectful, and professional.

"""
while True:
    question=input("\nYou : ")
    if question.lower()=="exit":
        break
    prompt=f"""
{system_prompt}
Student:
{question}
"""
    response=client.models.generate_content(
        model="models/gemini-3.1-flash-lite",
        contents=prompt
    )
    print("\nAssistant:")
    print(response.text)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset used by this dashboard
df = pd.read_csv("student_data.csv")

# Seaborn style
sns.set_theme(style="whitegrid")

# Create dashboard
fig, ax = plt.subplots(3, 2, figsize=(15, 14))

avg_marks = df.groupby("Department", as_index=False)["Marks"].mean()

sns.barplot(
    data=avg_marks,
    x="Department",
    y="Marks",
    ax=ax[0, 0]
 )

ax[0, 0].set_title("Average Marks by Department")
ax[0, 0].set_xlabel("Department")
ax[0, 0].set_ylabel("Average Marks")


# 2. Scatter Plot - Attendance vs Marks
sns.scatterplot(
    data=df,
    x="Attendance",
    y="Marks",
    ax=ax[0, 1]
 )

ax[0, 1].set_title("Attendance vs Marks")
ax[0, 1].set_xlabel("Attendance (%)")
ax[0, 1].set_ylabel("Marks")


# 3. Line Graph - Average Marks vs Attendance
attendance_marks = df.groupby(
    "Attendance",
    as_index=False
)["Marks"].mean()

sns.lineplot(
    data=attendance_marks,
    x="Attendance",
    y="Marks",
    marker="o",
    ax=ax[1, 0]
 )

ax[1, 0].set_title("Average Marks vs Attendance")
ax[1, 0].set_xlabel("Attendance (%)")
ax[1, 0].set_ylabel("Average Marks")


# 4. Pie Chart - City Distribution
city_count = df["City"].value_counts()

ax[1, 1].pie(
    city_count.values,
    labels=city_count.index,
    autopct="%1.1f%%"
 )

ax[1, 1].set_title("Student Distribution by City")


# 5. Histogram - Marks Distribution
sns.histplot(
    data=df,
    x="Marks",
    bins=5,
    ax=ax[2, 0]
 )

ax[2, 0].set_title("Marks Distribution")
ax[2, 0].set_xlabel("Marks")
ax[2, 0].set_ylabel("Number of Students")


# 6. Bar Chart - Students by Department
dept_count = df["Department"].value_counts().reset_index()

dept_count.columns = ["Department", "Students"]

sns.barplot(
    data=dept_count,
    x="Department",
    y="Students",
    ax=ax[2, 1]
 )

ax[2, 1].set_title("Students by Department")
ax[2, 1].set_xlabel("Department")
ax[2, 1].set_ylabel("Number of Students")


# Main Dashboard Title
fig.suptitle(
    "Student Performance Dashboard",
    fontsize=20
)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("student_performance.csv")

fig, ax = plt.subplots(2, 2, figsize=(13, 10))

ax[0, 0].hist(df["Attendance"], bins=8, edgecolor="black", label="Students")
ax[0, 0].set_title("Attendance Distribution")
ax[0, 0].set_xlabel("Attendance (%)")
ax[0, 0].set_ylabel("Number of Students")
ax[0, 0].legend()
ax[0, 0].grid(True, alpha=0.3)

ax[0, 1].scatter(df["Study_Hours"], df["Final_Marks"], label="Students")
ax[0, 1].set_title("Study Hours vs Final Marks")
ax[0, 1].set_xlabel("Study Hours")
ax[0, 1].set_ylabel("Final Marks")
ax[0, 1].legend()
ax[0, 1].grid(True, alpha=0.3)

avg_gender = df.groupby("Gender")["Final_Marks"].mean()

ax[1, 0].bar(avg_gender.index, avg_gender.values, label="Average Final Marks")
ax[1, 0].set_title("Average Final Marks by Gender")
ax[1, 0].set_xlabel("Gender")
ax[1, 0].set_ylabel("Average Final Marks")
ax[1, 0].legend()
ax[1, 0].grid(True, axis="y", alpha=0.3)

ax[1, 1].scatter(df["Attendance"], df["Final_Marks"], label="Students")
ax[1, 1].set_title("Attendance vs Final Marks")
ax[1, 1].set_xlabel("Attendance (%)")
ax[1, 1].set_ylabel("Final Marks")
ax[1, 1].legend()
ax[1, 1].grid(True, alpha=0.3)

fig.suptitle("Student Performance Analysis", fontsize=18)
plt.tight_layout()
plt.show()
```
